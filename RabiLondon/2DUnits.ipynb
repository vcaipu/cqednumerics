```python
import numpy as np
```

```python
# Of a COOPER PAIR, in SI UNITS
M_CP = 1.82e-30 #kg
Q_CP = -3.204e-19 #Coulombs

# Fuandmental Constants, in SI UNITS
mu_0 = 1.25663706212e-6 #Henry per meter / Newton per Ampere squared
epsilon_0 = 8.8541878128e-12 #Farads per meter
h_bar = 1.0545718e-34 #Joules seconds

# Fundamental constants object
fco = {
    "M_CP": M_CP,
    "Q_CP": Q_CP,
    "mu_0": mu_0,
    "epsilon_0": epsilon_0,
    "h_bar": h_bar
}
```

## Finding the charge density of Superconducting Alumnium

```python
# n_s from \xi

def n_sfromxi(xi,fco):
    m,q,mu_0,epsilon_0,hbar = fco["M_CP"],fco["Q_CP"],fco["mu_0"],fco["epsilon_0"],fco["h_bar"]
    c = 1/np.sqrt(mu_0*epsilon_0)
    h = h_bar * 2 * np.pi

    lambda_c = h / (m*c)
    n_s = m/(mu_0*q**2) * (lambda_c / (4*np.pi *xi**2))**2
    return n_s

def n_sfromlondon(lambda_L,fco):
    m,q,mu_0,epsilon_0,hbar = fco["M_CP"],fco["Q_CP"],fco["mu_0"],fco["epsilon_0"],fco["h_bar"]

    n_s = m / (mu_0 * q**2 * lambda_L**2)
    return n_s

def london_from_n_s(n_s,fco):
    m,q,mu_0,epsilon_0,hbar = fco["M_CP"],fco["Q_CP"],fco["mu_0"],fco["epsilon_0"],fco["h_bar"]
 
    return np.sqrt(m/(mu_0 * n_s * q**2))

xi = 0.39e-10
lambda_L = 1.6e-8 # 16nm
n_s_fromxi_res = n_sfromxi(xi,fco)
n_s_fromlondon_res = n_sfromlondon(lambda_L,fco)
print(f"n_s from xi {n_s_fromxi_res}")
print(f"n_s from london {n_s_fromlondon_res}")
print(f"London from xi n_s: {london_from_n_s(n_s_fromxi_res,fco)*1e9} nm")
print(f"n_s xi^3 {n_s_fromxi_res*xi**3}")
```

## Find the Right $\Delta Z$

Get $n_s$ from healing length, then Delta Z from n_s

```python
# N and Normalized Area from geometry and material
N = 100
area = 800


# Or compute a Delta Z from material n_s

n_s = 1.8e29 / 2
n_s = 5.68e28

def xifromn_s(n_s,fco):
    m,q,mu_0,epsilon_0,hbar = fco["M_CP"],fco["Q_CP"],fco["mu_0"],fco["epsilon_0"],fco["h_bar"]
    c = 1/np.sqrt(mu_0*epsilon_0)
    h = h_bar * 2 * np.pi

    lambda_c = h / (m*c)
    lambda_L = np.sqrt(m/(mu_0*n_s*q**2))
    return np.sqrt(lambda_c * lambda_L / (4*np.pi))

xi = xifromn_s(n_s,fco)
print(f"n_s: {n_s} / m^3 | xi: {xi*1e10} Angstroms | n_s xi^3: {n_s*xi**3}")

def compute_dz(n_s,N,area,fco):
    m,q,mu_0,epsilon_0,hbar = fco["M_CP"],fco["Q_CP"],fco["mu_0"],fco["epsilon_0"],fco["h_bar"]
    c = 1/np.sqrt(mu_0*epsilon_0)
    h = h_bar * 2 * np.pi

    lambda_c = h / (m*c)
    res_dz = (N/area) * (mu_0*q**2/m)**(3/4) * (4*np.pi/lambda_c)**(3/2) * (1/n_s)**(1/4)
    return res_dz

deltaZ = compute_dz(n_s,N,area,fco)

# Compute value for n_s
def compute_values(deltaZ,N,area,fco):
    m,q,mu_0,epsilon_0,hbar = fco["M_CP"],fco["Q_CP"],fco["mu_0"],fco["epsilon_0"],fco["h_bar"]
    c = 1/np.sqrt(mu_0*epsilon_0)
    h = h_bar * 2 * np.pi

    lambda_c = h / (m*c)

    n_s = ((N/area)**4) * (mu_0*q**2/m)**3 * (4*np.pi/lambda_c)**6 * (1/(deltaZ**4))
    lambda_L = np.sqrt(m/(mu_0*n_s*q**2))
    xi = np.sqrt(lambda_L * lambda_c / (4*np.pi))
    c_bar = c / xi
    return n_s, lambda_L, lambda_c, xi, c_bar

    # Debug
    # print(f"Speed of light: {c} m/s | Should be 299792458 m/s")
    # print(f"Compton wavelength: {lambda_c} m | Should be 1.21e-12 m")

n_s,lambda_L,lambda_C,xi,c_bar = compute_values(deltaZ,N,area,fco)
kappa = lambda_L / xi
print(f"n_s: {n_s} / m^3")
print(f"xi: {xi} m, {xi*1e10} Angstroms")
print(f"Material parameter n_s xi^3: {n_s*xi**3}")
print(f"lambda_L: {lambda_L} m | Kappa: {kappa}\n\n")

print(f"c_bar: {c_bar}| Kappa: {kappa}")
```
